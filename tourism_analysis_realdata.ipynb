# Japan Inbound Tourism Data Analysis

This project uses tourism statistics collected for my master's research to analyze changes in the geographic concentration and consumption of Chinese visitors to Japan.

The source workbook is `データ統計.xlsx`.

**Tech Stack:** Python · Pandas · Matplotlib · Jupyter Notebook


```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

FILE_PATH = Path("データ統計.xlsx")

if not FILE_PATH.exists():
    raise FileNotFoundError(
        "Please place データ統計.xlsx in the same folder as this notebook."
    )

print("Data file:", FILE_PATH)
```

## 1. Annual HHI trend

The workbook contains annual HHI values that measure the geographic concentration of visitor destinations.

A higher HHI indicates stronger concentration, while a lower HHI indicates greater geographic dispersion.


```python
annual_hhi = pd.read_excel(
    FILE_PATH,
    sheet_name="訪問地HII指数",
    header=None,
    usecols="A:B",
    skiprows=162,
    nrows=15
)

annual_hhi.columns = ["year", "hhi"]

annual_hhi["year"] = pd.to_numeric(annual_hhi["year"], errors="coerce")
annual_hhi["hhi"] = pd.to_numeric(annual_hhi["hhi"], errors="coerce")

annual_hhi = annual_hhi.dropna(subset=["year", "hhi"]).copy()
annual_hhi["year"] = annual_hhi["year"].astype(int)

annual_hhi
```

```python
plt.figure(figsize=(10, 5))
plt.plot(annual_hhi["year"], annual_hhi["hhi"], marker="o")
plt.title("Annual Geographic Concentration of Chinese Visitors to Japan")
plt.xlabel("Year")
plt.ylabel("HHI")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Quarterly HHI after the pandemic

Quarterly values are used to examine short-term and seasonal changes after inbound tourism recovered.


```python
quarterly_hhi = pd.read_excel(
    FILE_PATH,
    sheet_name="訪問地HII指数",
    header=None,
    usecols="D:E",
    skiprows=162,
    nrows=11
)

quarterly_hhi.columns = ["period", "hhi"]
quarterly_hhi["hhi"] = pd.to_numeric(quarterly_hhi["hhi"], errors="coerce")
quarterly_hhi = quarterly_hhi.dropna(subset=["period", "hhi"]).copy()

quarterly_hhi
```

```python
plt.figure(figsize=(11, 5))
plt.plot(quarterly_hhi["period"], quarterly_hhi["hhi"], marker="o")
plt.title("Quarterly HHI of Chinese Visitors to Japan")
plt.xlabel("Quarter")
plt.ylabel("HHI")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Total tourism consumption

The workbook also contains annual total spending by Chinese visitors to Japan.

Missing values during the pandemic period are retained as missing observations rather than being artificially filled.


```python
spending = pd.read_excel(
    FILE_PATH,
    sheet_name="各年度の消費総額（億円）",
    header=None,
    usecols="C:D",
    skiprows=3,
    nrows=14
)

spending.columns = ["year", "total_spending_100m_yen"]

spending["year"] = pd.to_numeric(spending["year"], errors="coerce")
spending["total_spending_100m_yen"] = pd.to_numeric(
    spending["total_spending_100m_yen"],
    errors="coerce"
)

spending = spending.dropna(subset=["year"]).copy()
spending["year"] = spending["year"].astype(int)

spending
```

```python
plot_spending = spending.dropna(subset=["total_spending_100m_yen"])

plt.figure(figsize=(10, 5))
plt.plot(
    plot_spending["year"],
    plot_spending["total_spending_100m_yen"],
    marker="o"
)
plt.title("Total Spending by Chinese Visitors to Japan")
plt.xlabel("Year")
plt.ylabel("Total Spending (100 million JPY)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. Simple comparison: 2019 vs 2024

This section provides a concise pre-pandemic/post-pandemic comparison.


```python
comparison = spending[
    spending["year"].isin([2019, 2024])
][["year", "total_spending_100m_yen"]].copy()

comparison["change_vs_2019_pct"] = (
    comparison["total_spending_100m_yen"]
    / comparison.loc[
        comparison["year"].eq(2019),
        "total_spending_100m_yen"
    ].iloc[0]
    - 1
) * 100

comparison
```
